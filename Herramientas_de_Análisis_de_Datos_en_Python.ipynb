```python
import pandas as pd
import os
from datetime import datetime
archivo = "gastos.csv"
costo_base = 15
fecha = datetime.now().strftime("%Y-%m-%d")
# Crear archivo si no existe
if not os.path.exists(archivo):
    df = pd.DataFrame(columns=["Fecha", "Categoria", "Descripcion", "Monto"])
    df.to_csv(archivo, index=False)
# Cargar datos
df = pd.read_csv(archivo)
# Entrada
viajes = int(input("¿Cuántos viajes vas a hacer hoy?: "))
total_dia = 0
print("\nDetalle de viajes:")
for i in range(1, viajes + 1):

    if i % 3 == 1:
        costo = costo_base
    elif i % 3 == 2:
        costo = 0
    else:
        costo = costo_base / 3
    total_dia += costo
    print(f"Viaje {i}: ${costo:.2f}")
    nuevo_gasto = {
        "Fecha": fecha,
        "Categoria": "Transporte",
        "Descripcion": f"Viaje {i}",
        "Monto": costo
    }
    df = pd.concat([df, pd.DataFrame([nuevo_gasto])], ignore_index=True)
# Guardar cambios
df.to_csv(archivo, index=False)
print("\n Transporte registrado automáticamente")
print(f" Total de viajes: {viajes}")
print(f" Gasto total del día: ${total_dia:.2f}")

```
