```python
import pandas as pd
import os
archivo = "gastos.csv"
# Crear archivo si no existe
if not os.path.exists(archivo):
    df = pd.DataFrame(columns=["Fecha", "Categoria", "Descripcion", "Monto"])
    df.to_csv(archivo, index=False)
# Cargar datos
df = pd.read_csv(archivo)
# Entrada de datos
fecha = input("Fecha (YYYY-MM-DD): ")
categoria = input("Categoría (Comida, Transporte, Escuela, Otros): ")
descripcion = input("Descripción: ")
monto = float(input("Monto: "))
# Nuevo gasto
nuevo_gasto = {
    "Fecha": fecha,
    "Categoria": categoria,
    "Descripcion": descripcion,
    "Monto": monto
}
# Agregar al DataFrame
df = pd.concat([df, pd.DataFrame([nuevo_gasto])], ignore_index=True)
# Guardar cambios
df.to_csv(archivo, index=False)
total = df["Monto"].sum()
por_categoria = df.groupby("Categoria")["Monto"].sum()
print("\nGasto registrado correctamente\n")
print(f"Total gastado: ${total:.2f}\n")
print("Gasto por categoría:")
print(por_categoria)
```
