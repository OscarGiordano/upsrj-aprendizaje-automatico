Este bloque de código importa varias librerías necesarias para trabajar con un modelo de Machine Learning.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.pipeline import Pipeline
```

Este bloque genera un conjunto de datos artificial para clasificación con 500 muestras y 2 características, donde ambas variables son informativas y no hay datos redundantes. También se define un solo grupo por clase y se usa random_state=42 para obtener siempre los mismos resultados. Los datos se guardan en X (características) y y (clases), y luego se imprimen.

```python
# 1. tenemos una base de datos con 500 muestras
# 2. tenemos dos features
# 3. no hay features redundantes
# 4. numero de features informativos
X, y = make_classification(
    n_samples=500,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)
print(X)
print(y)
```

```python
# Dividimos datos para entrenamiento y pruebas
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
print(X_train)
print(X_test)
```

Este bloque normaliza los datos usando StandardScaler para que las variables tengan valores en la misma escala. Primero, el método fit_transform ajusta el escalador con los datos de entrenamiento y los transforma. Después, transform aplica esa misma escala a los datos de prueba para mantener consistencia. Finalmente, se imprimen los datos escalados de entrenamiento y prueba.

```python
# Escalamos datos para procesar (preprocesamiento)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)
print(X_train)
print(X_test)
```

generamos el modelo para alimentarlo con los datos

```python
# Generamos nuestro modelo y lo alimentamos
model = LogisticRegression()
model.fit(X_train, y_train)
```

aqui creamos los comandos para nuestra prediccion

```python
# Generamos prediccion
y_pred = model.predict(X_test)
print(y_pred)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion matrix:", confusion_matrix(y_test, y_pred))
print("Classification report:", classification_report(y_test, y_pred))
```

```python
def plot_decision_boundary(X, y, model):
  x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
  y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1
  xx, yy = np.meshgrid(
      np.linspace(x_min, x_max, 200),
      np.linspace(y_min, y_max, 200)
  )
  Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
  Z = Z.reshape(xx.shape)
  plt.contourf(xx, yy, Z, alpha=0.3)
  plt.scatter(X[:,0], X[:,1], c=y, edgecolors="k", s=50)
  plt.xlabel("Feature 1")
  plt.ylabel("Feature 2")
  plt.title("Logistic Regression Decision Boundary")
  plt.show()
```

```python
plot_decision_boundary(X_test, y_test, model)
```

PIPELINE
aqui generamos nuestro pipeline, escribi la libreria de pipeline al principio de este codigo para llevar un orden

```python

X, y = make_classification(
    n_samples=500,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)

# Dividir datos
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Pipeline con REGRESIÓN LOGÍSTICA
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression())
])

# Entrenar
pipeline.fit(X_train, y_train)

# Predicción
y_pred = pipeline.predict(X_test)

# Evaluación
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification report:\n", classification_report(y_test, y_pred))


# --------- Frontera de decisión ----------
def plot_decision_boundary(X, y, model):

    x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
    y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 200),
        np.linspace(y_min, y_max, 200)
    )

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X[:,0], X[:,1], c=y, edgecolors="k", s=50)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.title("Logistic Regression Decision Boundary")
    plt.show()

plot_decision_boundary(X_test, y_test, pipeline)
```
