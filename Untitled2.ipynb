# IMPORTACION DE LIBRERIAS
SE DECLARAN LAS LIBRERIAS NECESARIAS PARA ESTE CODIGO

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

# DEFINICIÓN DE LOS DATOS
Se define la relación matemática del modelo como y = m*x + b.
Se crea el arreglo X que representa la variable independiente,
el cual se reorganiza en forma de columna para que sea compatible
con el modelo de regresión lineal.
Se define el arreglo y como la variable dependiente.
Finalmente, se imprime el contenido de y para verificar los datos.

```python
# y = m * x + b
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
# variable independiente
y = np.array([2, 4, 5, 4, 5])
print(y)
```

# GRAFICACION DE LOS DATOS
Se genera una gráfica de dispersión utilizando los valores de X y y.
Se asigna la etiqueta "X" al eje horizontal y "Y" al eje vertical.
Se establece un título para identificar la gráfica como los datos originales.
Se activa la cuadrícula para facilitar la interpretación visual.
Finalmente, se muestra la gráfica en pantalla.

```python
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Datos originales")
plt.grid(True)
plt.show()
```

# CREACIÓN Y ENTRENAMIENTO DEL MODELO
Se crea el modelo de regresión lineal.
Se ajusta el modelo utilizando los datos de la variable independiente X
y la variable dependiente y.
Se obtienen los parámetros del modelo: la pendiente m (coeficiente)
y el intercepto b (sesgo).
Finalmente, se imprimen los valores de la pendiente y el intercepto
con dos cifras decimales para una mejor presentación.


```python
# Generamos nuestro modelo de regresion linear
model = LinearRegression()
# Alineamos nuestro modelo
model.fit(X, y)
# parametros: m (peso), b (sesgo)
m = model.coef_[0]
b = model.intercept_
print(f"Pendiente(m):{m:.2f}")
print(f"intercepto(b):{b:.2f}")
```

# PREDICCIÓN Y EVALUACIÓN DEL MODELO
Se generan las predicciones del modelo de regresión lineal utilizando
los valores de la variable independiente X.
Se evalúa el desempeño del modelo comparando los valores reales y
los valores predichos mediante el error cuadrático medio


```python
#generamos una predicción
y_pred = model.predict(X)
#Evaluamos esa predicción
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y,y_pred)
print(f"Mean Squared Error:{mse:2f}")
print(f"R2 Score:{r2:2f}")
```

# VISUALIZACIÓN DEL MODELO DE REGRESIÓN
Se genera una gráfica de dispersión con los datos reales utilizando X y y.
Se traza la línea del modelo de regresión lineal a partir de las
predicciones obtenidas.
Finalmente, se muestra la gráfica en pantalla.


```python
plt.scatter(X,y,label="Datos reales")
plt.plot(X,y_pred,color="red",label="Modelo lineal")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Regresion Lineal simple")
plt.grid(True)
plt.show()
```
